```python
import zipfile
import os

zip_file_path = '/content/Electric_Vehicle_Population_Data.csv.zip'
output_dir = '/content/'

with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(output_dir)

print(f"Extracted all files to {output_dir}")
print(os.listdir(output_dir))
```

```python
import pandas as pd

df = pd.read_csv('/content/Electric_Vehicle_Population_Data.csv')

df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
df = df.drop_duplicates()

print("Data Cleaned Successfully")
```

```python
print("Total EV Vehicles:", len(df))
```

```python
df['Make'].value_counts().head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df['Make'].value_counts().head(10).plot(kind='bar', figsize=(10,5))

plt.title('Top 10 EV Manufacturers')
plt.xlabel('Company')
plt.ylabel('Number of Vehicles')

plt.show()
```

```python
df['Electric Vehicle Type'].value_counts()
```

```python
df['Electric Vehicle Type'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(7,7)
)

plt.title('Electric Vehicle Type Distribution')
plt.ylabel('')

plt.show()
```

```python
df['Model Year'].value_counts().sort_index().plot(
    figsize=(12,5)
)

plt.title('Growth of Electric Vehicles by Year')
plt.xlabel('Year')
plt.ylabel('Number of Vehicles')

plt.show()
```

```python
import seaborn as sns
df['City'].value_counts().head(10).plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Top 10 Cities with EVs')
plt.xlabel('City')
plt.ylabel('Vehicle Count')

plt.show()
```

```python
numeric_df = df.select_dtypes(include=['int64', 'float64'])

sns.heatmap(numeric_df.corr(), annot=True)

plt.title('Correlation Heatmap')

plt.show()
```
